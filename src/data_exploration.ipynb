```python
import pandas as pd
import os
```

```python
# Feature names extracted from the unsw research paper
feature_names = ['scrip', 'sport', 'dstip', 'dsport', 'proto', 'state', 'dur', 'sbytes', 'dbytes', 'sttl', 'dttl', 'sloss', 'dloss', 'service', 'sload', 'dload', 'spkts', 'dpkts', 'swin', 'dwin', 'stcpb', 'dtcpb', 'smeans', 'dmeansz', 'trans_depth', 'res_bdy_len', 'sjit', 'djit', 'stime', 'ltime', 'sintpkt', 'dintpkt', 'tcprtt', 'synack', 'ackdat', 'is_sm_ips_ports', 'ct_state_ttl', 'ct_flw_http_mthd', 'is_ftp_login', 'ct_ftp_cmd', 'ct_srv_src', 'ct_srv_dst', 'ct_dst_ltm', 'ct_src_ ltm', 'ct_src_dport_ltm', 'ct_dst_sport_ltm', 'ct_dst_src_ltm', 'attack_cat', 'label']

def load_unsw(path: str) -> pd.DataFrame | None:
    datasets_container = []
    for file in os.listdir(path):
        if file.endswith('.csv') and "UNSW" in file:
            dataset = pd.read_csv(os.path.join(path, file), names=feature_names)
            datasets_container.append(dataset)

    try:
        return pd.concat(datasets_container, ignore_index=True)
    except:
        print("Error: Unable to concatenate datasets. Please check the files in the directory.")
        return None

```

```python
def load_cic(path: str) -> pd.DataFrame | None:
    datasets_container = []
    for file in os.listdir(path):
        if file.endswith('.csv') and "pcap_ISCX" in file:
            data = pd.read_csv(os.path.join(path, file))
            datasets_container.append(data)

    try:
        return pd.concat(datasets_container, ignore_index=True)
    except:
        print("Error: Unable to concatenate datasets. Please check the files in the directory.")
        return None
```

```python
cic_dataset = load_cic("./datasets/cic")
```
